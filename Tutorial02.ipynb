# Question 1 :  ⁠Create a numpy array of size 2 x 2

```python
import numpy as np
array1 = np.array([[1, 4],
                   [5, 7]])
print(array1)
```

# Question 2 : Create a random numpy array of size 3 x 4

```python
array2 = np.random.rand(3, 4)
print(array2)
```

# Question 3 : Create a numpy array of ones and a numpy array of zeroes

```python
array_ones = np.ones((2, 2))

array_zeroes = np.zeros((2, 2))

print("Array of ones:")
print(array_ones)

print("array of zeroes:")
print(array_zeroes)
```

# Question 4 : Create a random numpy array of size 6x3 and transpose it

```python
array4 = np.random.rand(2, 3)

print("original array:")
print(array4)

transposed_array = array4.T

print("transposed array:")
print(transposed_array)
```

# Question 5 : Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n = 2
p = 3
m = 4

A = np.random.rand(n, p)
B = np.random.rand(p, m)

print("Matrix A")
print(A)
print("Matrix B")
print(B)

result = np.dot(A, B)

print("dot product:")
print(result)
```
